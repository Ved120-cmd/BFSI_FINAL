```python
"""
Insurance Data Visualization - SIMPLIFIED VERSION
==================================================
Easy-to-run plotting code with error handling
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

# Set style for better-looking plots
plt.style.use('default')
sns.set_palette("husl")
plt.rcParams['figure.dpi'] = 100
plt.rcParams['savefig.dpi'] = 300

print("="*80)
print("INSURANCE DATA VISUALIZATION - SIMPLIFIED VERSION")
print("="*80)
```

```python
try:
    print("\n📂 Loading ORIGINAL datasets...")
    customers_orig = pd.read_csv('../needed/insurance_dataset.csv')
    claims_orig = pd.read_csv('../needed/insurance_claims.csv')
    print(f"✅ Original Customers loaded: {customers_orig.shape}")
    print(f"✅ Original Claims loaded: {claims_orig.shape}")
    
    print("\n📋 Customer Dataset Columns:")
    print(customers_orig.columns.tolist())
    print("\n📋 Claims Dataset Columns (first 20):")
    print(claims_orig.columns.tolist()[:20])
    
except Exception as e:
    print(f"❌ Error loading data: {e}")
    exit(1)
```

```python
def save_plot(filename):
    """Helper function to save plots"""
    filepath = f'../outputs/original_{filename}'
    plt.tight_layout()
    plt.savefig(filepath, dpi=300, bbox_inches='tight')
    print(f"  ✅ Saved: original_{filename}")
    plt.close()
```

```python
print("\n📊 Generating Plot 1: Customer Overview Dashboard...")
try:
    fig, axes = plt.subplots(2, 3, figsize=(16, 10))
    fig.suptitle('ORIGINAL CUSTOMER DATA - OVERVIEW DASHBOARD', fontsize=18, fontweight='bold', y=0.995)
    
    # 1. Age Distribution
    axes[0, 0].hist(customers_orig['Age'], bins=30, color='steelblue', edgecolor='black', alpha=0.7)
    axes[0, 0].axvline(customers_orig['Age'].mean(), color='red', linestyle='--', linewidth=2, 
                       label=f'Mean: {customers_orig["Age"].mean():.1f} years')
    axes[0, 0].set_xlabel('Age (years)', fontsize=11, fontweight='bold')
    axes[0, 0].set_ylabel('Number of Customers', fontsize=11, fontweight='bold')
    axes[0, 0].set_title('Age Distribution', fontsize=13, fontweight='bold')
    axes[0, 0].legend()
    axes[0, 0].grid(alpha=0.3)
    
    # 2. Gender Distribution
    gender_counts = customers_orig['Gender'].value_counts()
    colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#ff99cc']
    axes[0, 1].pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', 
                   startangle=90, colors=colors[:len(gender_counts)])
    axes[0, 1].set_title('Gender Distribution', fontsize=13, fontweight='bold')
    
    # 3. Policy Type Distribution
    policy_counts = customers_orig['Policy_Type'].value_counts()
    axes[0, 2].bar(policy_counts.index, policy_counts.values, color='coral', edgecolor='black', alpha=0.7)
    axes[0, 2].set_xlabel('Policy Type', fontsize=11, fontweight='bold')
    axes[0, 2].set_ylabel('Number of Customers', fontsize=11, fontweight='bold')
    axes[0, 2].set_title('Policy Type Distribution', fontsize=13, fontweight='bold')
    axes[0, 2].grid(axis='y', alpha=0.3)
    
    # 4. Regional Distribution
    region_counts = customers_orig['Region'].value_counts()
    axes[1, 0].barh(region_counts.index, region_counts.values, color='mediumseagreen', edgecolor='black', alpha=0.7)
    axes[1, 0].set_xlabel('Number of Customers', fontsize=11, fontweight='bold')
    axes[1, 0].set_ylabel('Region', fontsize=11, fontweight='bold')
    axes[1, 0].set_title('Regional Distribution', fontsize=13, fontweight='bold')
    axes[1, 0].grid(axis='x', alpha=0.3)
    
    # 5. Premium Distribution
    axes[1, 1].hist(customers_orig['Premium'].dropna(), bins=30, color='orange', edgecolor='black', alpha=0.7)
    axes[1, 1].axvline(customers_orig['Premium'].median(), color='darkred', linestyle='--', linewidth=2, 
                       label=f'Median: ₹{customers_orig["Premium"].median():.2f}')
    axes[1, 1].set_xlabel('Premium (₹)', fontsize=11, fontweight='bold')
    axes[1, 1].set_ylabel('Number of Customers', fontsize=11, fontweight='bold')
    axes[1, 1].set_title('Premium Distribution', fontsize=13, fontweight='bold')
    axes[1, 1].legend()
    axes[1, 1].grid(alpha=0.3)
    
    # 6. Claim Count Distribution
    claim_counts = customers_orig['Claim_Count'].value_counts().sort_index()
    axes[1, 2].bar(claim_counts.index, claim_counts.values, color='purple', edgecolor='black', alpha=0.7)
    axes[1, 2].set_xlabel('Number of Claims', fontsize=11, fontweight='bold')
    axes[1, 2].set_ylabel('Number of Customers', fontsize=11, fontweight='bold')
    axes[1, 2].set_title('Claim Count Distribution', fontsize=13, fontweight='bold')
    axes[1, 2].grid(axis='y', alpha=0.3)
    
    save_plot('01_customer_overview.png')
except Exception as e:
    print(f"  ❌ Error: {e}")
```

```python
print("\n📊 Generating Plot 2: Customer Satisfaction Analysis...")
try:
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle('ORIGINAL DATA - CUSTOMER SATISFACTION ANALYSIS', fontsize=18, fontweight='bold', y=0.995)
    
    # 1. Satisfaction Distribution
    axes[0, 0].hist(customers_orig['Customer_Satisfaction'].dropna(), bins=10, 
                    color='lightcoral', edgecolor='black', alpha=0.7)
    axes[0, 0].axvline(customers_orig['Customer_Satisfaction'].mean(), color='darkred', 
                       linestyle='--', linewidth=2, 
                       label=f'Mean: {customers_orig["Customer_Satisfaction"].mean():.2f}/10')
    axes[0, 0].set_xlabel('Satisfaction Score (1-10)', fontsize=11, fontweight='bold')
    axes[0, 0].set_ylabel('Number of Customers', fontsize=11, fontweight='bold')
    axes[0, 0].set_title('Overall Satisfaction Distribution', fontsize=13, fontweight='bold')
    axes[0, 0].legend()
    axes[0, 0].grid(alpha=0.3)
    
    # 2. Satisfaction by Policy Type
    policy_sat = customers_orig.groupby('Policy_Type')['Customer_Satisfaction'].mean().sort_values()
    axes[0, 1].barh(policy_sat.index, policy_sat.values, color='teal', edgecolor='black', alpha=0.7)
    axes[0, 1].set_xlabel('Average Satisfaction Score', fontsize=11, fontweight='bold')
    axes[0, 1].set_ylabel('Policy Type', fontsize=11, fontweight='bold')
    axes[0, 1].set_title('Avg Satisfaction by Policy Type', fontsize=13, fontweight='bold')
    axes[0, 1].grid(axis='x', alpha=0.3)
    
    # 3. Satisfaction by Region
    region_sat = customers_orig.groupby('Region')['Customer_Satisfaction'].mean().sort_values(ascending=False)
    axes[1, 0].bar(region_sat.index, region_sat.values, color='dodgerblue', edgecolor='black', alpha=0.7)
    axes[1, 0].set_xlabel('Region', fontsize=11, fontweight='bold')
    axes[1, 0].set_ylabel('Average Satisfaction', fontsize=11, fontweight='bold')
    axes[1, 0].set_title('Avg Satisfaction by Region', fontsize=13, fontweight='bold')
    axes[1, 0].grid(axis='y', alpha=0.3)
    
    # 4. Satisfaction vs Claims (Scatter)
    axes[1, 1].scatter(customers_orig['Claim_Count'], customers_orig['Customer_Satisfaction'], 
                       alpha=0.5, color='indianred', edgecolors='black', s=50)
    axes[1, 1].set_xlabel('Number of Claims', fontsize=11, fontweight='bold')
    axes[1, 1].set_ylabel('Satisfaction Score', fontsize=11, fontweight='bold')
    axes[1, 1].set_title('Satisfaction vs Claims', fontsize=13, fontweight='bold')
    axes[1, 1].grid(alpha=0.3)
    
    # Add correlation
    valid_data = customers_orig[['Claim_Count', 'Customer_Satisfaction']].dropna()
    if len(valid_data) > 0:
        corr = valid_data.corr().iloc[0, 1]
        axes[1, 1].text(0.05, 0.95, f'Correlation: {corr:.3f}', transform=axes[1, 1].transAxes,
                       fontsize=11, verticalalignment='top', 
                       bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    
    save_plot('02_satisfaction_analysis.png')
except Exception as e:
    print(f"  ❌ Error: {e}")
```

```python
print("\n📊 Generating Plot 3: Premium & Claims Analysis...")
try:
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle('ORIGINAL DATA - PREMIUM & CLAIMS ANALYSIS', fontsize=18, fontweight='bold', y=0.995)
    
    # 1. Premium by Policy Type
    policy_prem = customers_orig.groupby('Policy_Type')['Premium'].mean().sort_values(ascending=False)
    axes[0, 0].bar(policy_prem.index, policy_prem.values, color='goldenrod', edgecolor='black', alpha=0.7)
    axes[0, 0].set_xlabel('Policy Type', fontsize=11, fontweight='bold')
    axes[0, 0].set_ylabel('Average Premium (₹)', fontsize=11, fontweight='bold')
    axes[0, 0].set_title('Avg Premium by Policy Type', fontsize=13, fontweight='bold')
    axes[0, 0].grid(axis='y', alpha=0.3)
    
    # 2. Premium by Region
    region_prem = customers_orig.groupby('Region')['Premium'].mean().sort_values(ascending=False)
    axes[0, 1].barh(region_prem.index, region_prem.values, color='mediumorchid', edgecolor='black', alpha=0.7)
    axes[0, 1].set_xlabel('Average Premium (₹)', fontsize=11, fontweight='bold')
    axes[0, 1].set_ylabel('Region', fontsize=11, fontweight='bold')
    axes[0, 1].set_title('Avg Premium by Region', fontsize=13, fontweight='bold')
    axes[0, 1].grid(axis='x', alpha=0.3)
    
    # 3. Claims by Policy Type
    policy_claims = customers_orig.groupby('Policy_Type')['Claim_Count'].mean().sort_values(ascending=False)
    axes[1, 0].bar(policy_claims.index, policy_claims.values, color='crimson', edgecolor='black', alpha=0.7)
    axes[1, 0].set_xlabel('Policy Type', fontsize=11, fontweight='bold')
    axes[1, 0].set_ylabel('Average Claims', fontsize=11, fontweight='bold')
    axes[1, 0].set_title('Avg Claims by Policy Type', fontsize=13, fontweight='bold')
    axes[1, 0].grid(axis='y', alpha=0.3)
    
    # 4. Age vs Premium (Scatter)
    axes[1, 1].scatter(customers_orig['Age'], customers_orig['Premium'], 
                       alpha=0.5, c=customers_orig['Claim_Count'], cmap='YlOrRd', 
                       s=50, edgecolors='black')
    axes[1, 1].set_xlabel('Age (years)', fontsize=11, fontweight='bold')
    axes[1, 1].set_ylabel('Premium (₹)', fontsize=11, fontweight='bold')
    axes[1, 1].set_title('Age vs Premium (Color = Claims)', fontsize=13, fontweight='bold')
    axes[1, 1].grid(alpha=0.3)
    cbar = plt.colorbar(axes[1, 1].collections[0], ax=axes[1, 1])
    cbar.set_label('Claim Count', fontsize=10)
    
    save_plot('03_premium_claims_analysis.png')
except Exception as e:
    print(f"  ❌ Error: {e}")
```

```python
print("\n📊 Generating Plot 4: Customer Acquisition Trends...")
try:
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle('ORIGINAL DATA - CUSTOMER ACQUISITION TRENDS', fontsize=18, fontweight='bold', y=0.995)
    
    # Convert Date_Joined to datetime
    customers_orig['Date_Joined_dt'] = pd.to_datetime(customers_orig['Date_Joined'])
    customers_orig['Join_Year'] = customers_orig['Date_Joined_dt'].dt.year
    customers_orig['Join_Month'] = customers_orig['Date_Joined_dt'].dt.month
    customers_orig['Join_Quarter'] = customers_orig['Date_Joined_dt'].dt.quarter
    
    # 1. Customers by Join Year
    year_counts = customers_orig['Join_Year'].value_counts().sort_index()
    axes[0, 0].plot(year_counts.index, year_counts.values, marker='o', linewidth=2, 
                    markersize=8, color='steelblue')
    axes[0, 0].fill_between(year_counts.index, year_counts.values, alpha=0.3, color='steelblue')
    axes[0, 0].set_xlabel('Year', fontsize=11, fontweight='bold')
    axes[0, 0].set_ylabel('New Customers', fontsize=11, fontweight='bold')
    axes[0, 0].set_title('Customer Acquisition by Year', fontsize=13, fontweight='bold')
    axes[0, 0].grid(alpha=0.3)
    
    # 2. Customers by Quarter (latest year)
    latest_year = customers_orig['Join_Year'].max()
    latest_year_data = customers_orig[customers_orig['Join_Year'] == latest_year]
    quarter_counts = latest_year_data['Join_Quarter'].value_counts().sort_index()
    axes[0, 1].bar(quarter_counts.index, quarter_counts.values, color='coral', edgecolor='black', alpha=0.7)
    axes[0, 1].set_xlabel('Quarter', fontsize=11, fontweight='bold')
    axes[0, 1].set_ylabel('New Customers', fontsize=11, fontweight='bold')
    axes[0, 1].set_title(f'Acquisitions by Quarter ({latest_year})', fontsize=13, fontweight='bold')
    axes[0, 1].grid(axis='y', alpha=0.3)
    
    # 3. Gender Distribution Over Time
    gender_year = customers_orig.groupby(['Join_Year', 'Gender']).size().unstack(fill_value=0)
    gender_year.plot(kind='bar', stacked=True, ax=axes[1, 0], alpha=0.7, edgecolor='black')
    axes[1, 0].set_xlabel('Year', fontsize=11, fontweight='bold')
    axes[1, 0].set_ylabel('Number of Customers', fontsize=11, fontweight='bold')
    axes[1, 0].set_title('Gender Distribution Over Time', fontsize=13, fontweight='bold')
    axes[1, 0].legend(title='Gender', fontsize=9)
    axes[1, 0].grid(axis='y', alpha=0.3)
    axes[1, 0].tick_params(axis='x', rotation=45)
    
    # 4. Policy Type Distribution Over Time
    policy_year = customers_orig.groupby(['Join_Year', 'Policy_Type']).size().unstack(fill_value=0)
    policy_year.plot(kind='line', marker='o', ax=axes[1, 1], linewidth=2, markersize=6)
    axes[1, 1].set_xlabel('Year', fontsize=11, fontweight='bold')
    axes[1, 1].set_ylabel('Number of Policies', fontsize=11, fontweight='bold')
    axes[1, 1].set_title('Policy Type Trends Over Time', fontsize=13, fontweight='bold')
    axes[1, 1].legend(title='Policy Type', fontsize=9)
    axes[1, 1].grid(alpha=0.3)
    
    save_plot('04_acquisition_trends.png')
except Exception as e:
    print(f"  ❌ Error: {e}")
```

```python
print("\n📊 Generating Plot 5: Claims - Fraud Analysis...")
try:
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle('ORIGINAL CLAIMS DATA - FRAUD ANALYSIS', fontsize=18, fontweight='bold', y=0.995)
    
    # 1. Fraud Distribution
    fraud_counts = claims_orig['fraud_reported'].value_counts()
    colors_fraud = ['lightgreen', 'crimson']
    labels_fraud = ['Legitimate (N)', 'Fraud (Y)']
    axes[0, 0].pie(fraud_counts, labels=labels_fraud, autopct='%1.1f%%', 
                   startangle=90, colors=colors_fraud, explode=(0, 0.1))
    axes[0, 0].set_title(f'Fraud Rate: {(fraud_counts.get("Y", 0) / fraud_counts.sum() * 100):.1f}%', 
                         fontsize=13, fontweight='bold')
    
    # 2. Fraud by Incident Type
    fraud_by_incident = claims_orig.groupby('incident_type')['fraud_reported'].apply(
        lambda x: (x == 'Y').sum() / len(x) * 100).sort_values(ascending=False)
    axes[0, 1].barh(fraud_by_incident.index, fraud_by_incident.values, 
                    color='orangered', edgecolor='black', alpha=0.7)
    axes[0, 1].set_xlabel('Fraud Rate (%)', fontsize=11, fontweight='bold')
    axes[0, 1].set_ylabel('Incident Type', fontsize=11, fontweight='bold')
    axes[0, 1].set_title('Fraud Rate by Incident Type', fontsize=13, fontweight='bold')
    axes[0, 1].grid(axis='x', alpha=0.3)
    
    # 3. Fraud by Severity
    fraud_by_severity = claims_orig.groupby('incident_severity')['fraud_reported'].apply(
        lambda x: (x == 'Y').sum() / len(x) * 100).sort_values(ascending=False)
    axes[1, 0].bar(fraud_by_severity.index, fraud_by_severity.values, 
                   color='darkred', edgecolor='black', alpha=0.7)
    axes[1, 0].set_xlabel('Severity Level', fontsize=11, fontweight='bold')
    axes[1, 0].set_ylabel('Fraud Rate (%)', fontsize=11, fontweight='bold')
    axes[1, 0].set_title('Fraud Rate by Severity', fontsize=13, fontweight='bold')
    axes[1, 0].tick_params(axis='x', rotation=45)
    axes[1, 0].grid(axis='y', alpha=0.3)
    
    # 4. Claim Amount: Fraud vs Legitimate
    fraud_mask = claims_orig['fraud_reported'] == 'Y'
    data_to_plot = [claims_orig[~fraud_mask]['total_claim_amount'], 
                    claims_orig[fraud_mask]['total_claim_amount']]
    bp = axes[1, 1].boxplot(data_to_plot, labels=['Legitimate', 'Fraud'],
                            patch_artist=True)
    bp['boxes'][0].set_facecolor('lightgreen')
    bp['boxes'][1].set_facecolor('crimson')
    for box in bp['boxes']:
        box.set_alpha(0.7)
    axes[1, 1].set_ylabel('Claim Amount ($)', fontsize=11, fontweight='bold')
    axes[1, 1].set_title('Claim Amount: Fraud vs Legitimate', fontsize=13, fontweight='bold')
    axes[1, 1].grid(axis='y', alpha=0.3)
    
    save_plot('05_fraud_analysis.png')
except Exception as e:
    print(f"  ❌ Error: {e}")
```

```python
print("\n📊 Generating Plot 7: Claims - Financial Analysis...")
try:
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle('ORIGINAL CLAIMS DATA - FINANCIAL ANALYSIS', fontsize=18, fontweight='bold', y=0.995)
    
    # 1. Total Claim Amount Distribution
    axes[0, 0].hist(claims_orig['total_claim_amount'], bins=50, 
                    color='goldenrod', edgecolor='black', alpha=0.7)
    axes[0, 0].axvline(claims_orig['total_claim_amount'].median(), color='darkred', 
                       linestyle='--', linewidth=2, 
                       label=f'Median: ${claims_orig["total_claim_amount"].median():.0f}')
    axes[0, 0].set_xlabel('Total Claim Amount ($)', fontsize=11, fontweight='bold')
    axes[0, 0].set_ylabel('Number of Claims', fontsize=11, fontweight='bold')
    axes[0, 0].set_title('Claim Amount Distribution', fontsize=13, fontweight='bold')
    axes[0, 0].legend()
    axes[0, 0].grid(alpha=0.3)
    
    # 2. Claim Components Breakdown
    claim_components = claims_orig[['injury_claim', 'property_claim', 'vehicle_claim']].mean()
    axes[0, 1].pie(claim_components, labels=['Injury', 'Property', 'Vehicle'], 
                   autopct='%1.1f%%', startangle=90, colors=['#ff9999', '#66b3ff', '#99ff99'])
    axes[0, 1].set_title('Average Claim Breakdown', fontsize=13, fontweight='bold')
    
    # 3. Policy Premium Distribution
    axes[1, 0].hist(claims_orig['policy_annual_premium'], bins=50, 
                    color='teal', edgecolor='black', alpha=0.7)
    axes[1, 0].axvline(claims_orig['policy_annual_premium'].median(), color='darkred', 
                       linestyle='--', linewidth=2, 
                       label=f'Median: ${claims_orig["policy_annual_premium"].median():.0f}')
    axes[1, 0].set_xlabel('Annual Premium ($)', fontsize=11, fontweight='bold')
    axes[1, 0].set_ylabel('Number of Policies', fontsize=11, fontweight='bold')
    axes[1, 0].set_title('Policy Premium Distribution', fontsize=13, fontweight='bold')
    axes[1, 0].legend()
    axes[1, 0].grid(alpha=0.3)
    
    # 4. Premium vs Claim Amount
    axes[1, 1].scatter(claims_orig['policy_annual_premium'], 
                       claims_orig['total_claim_amount'], 
                       alpha=0.5, c=(claims_orig['fraud_reported'] == 'Y'), 
                       cmap='RdYlGn_r', s=50, edgecolors='black')
    axes[1, 1].set_xlabel('Annual Premium ($)', fontsize=11, fontweight='bold')
    axes[1, 1].set_ylabel('Claim Amount ($)', fontsize=11, fontweight='bold')
    axes[1, 1].set_title('Premium vs Claim (Red = Fraud)', fontsize=13, fontweight='bold')
    axes[1, 1].grid(alpha=0.3)
    
    save_plot('07_financial_analysis.png')
except Exception as e:
    print(f"  ❌ Error: {e}")
```
